```python
import sqlite3
import pandas as pd

# built into Python
# for readable output

con = sqlite3.connect(r'..\data\raw\airtel.db') # two dots: notebooks/

def q(sql):
    return pd.read_sql(sql, con)

# the same helper as
# the IPL project
```

```python
q("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name;
""")
```

```python
q("""
SELECT 'circles' AS tbl, COUNT(*) AS n FROM circles
UNION ALL
SELECT 'plans', COUNT(*) FROM plans
UNION ALL
SELECT 'customers', COUNT(*) FROM customers
UNION ALL
SELECT 'usage_daily', COUNT(*) FROM usage_daily
UNION ALL
SELECT 'recharges', COUNT(*) FROM recharges
UNION ALL
SELECT 'complaints', COUNT(*) FROM complaints
ORDER BY tbl;
""")
```
